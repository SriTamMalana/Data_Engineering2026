```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
from pyspark.sql import SparkSession

emp = spark.read.csv(
    '/Volumes/workspace/default/vol1/emp_rec.csv',header = True,inferSchema=True
    )
dept = spark.read.format('csv').option('header',True).option('inferSchema',True).load('/Volumes/workspace/default/vol1/Dept_rec.csv')
#dept.display()

salgrade = spark.read.csv(
    '/Volumes/workspace/default/vol1/Salgrade_rec.csv',header=True, inferSchema=True
)
```

### 1. EACH DEPTNO WISE TOP 2 HIGHEST PAID EMPLOYEE NAME, SAL ?

```python
from pyspark.sql.window import * 
win = Window.partitionBy('deptno').orderBy(col('sal'))
emp01 = (emp.withColumn('drnk',dense_rank().over(win))
            .select('empno','ename','sal').filter(col('drnk') <= 2))
            #.display())
```

### 2. DISPLAY LOCATION WISE 2ND HIGHEST SALARY LOC, ENAME,SAL  FOR ALL LOCATION EVEN THERE IS NO EMPLOYEE

```python

win = Window.partitionBy('loc').orderBy(col('sal').desc())
emp02 = (emp.join(dept,on=[emp['deptno'] == dept['deptno']],how='inner')
         .withColumn('drnk',dense_rank().over(win))
         .select('loc','ename','sal').filter(col('drnk')==2).display())
```

### 3. DISPLAY EMPNO,ENAME,RANKING OF SALARY (USE DENSE_RANK),DISPLAY RANK IF 1 TO 3  THEN  AS "A GRADE", 4 TO 7 "B GRADE" ELSE OTHERS ?

```python
# emp.display()
emp_rank = ( emp.withColumn('drnk',dense_rank().over(
    Window.partitionBy().orderBy(col('sal').desc())
)) 
.withColumn('GRADE',when(col('drnk')<=3,lit('A Grade'))
.when(col('drnk').between(4,7) ,lit('B Grade'))
.otherwise(lit('Others'))).display()

)
```

### 4.DISPLAY DNAME , LOCATION ,ENAME,SAL FOR 3RD HIGHEST PAID EMPLOYEE?

```python
win_spec = Window.partitionBy().orderBy(col('sal').desc())
win = (emp.join(dept,'deptno','inner')
       .withColumn('drnk',dense_rank().over(win_spec))
       .filter(col('drnk')==3).select('dname','loc','ename','sal').display()
       )
```
